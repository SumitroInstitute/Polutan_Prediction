```python
import pandas as pd

# Path ke file CSV hasil akhir
csv_path = "/workspaces/Polutan_Prediction/Data_Final/Kelapa Gading Daily Averages.csv"

# Baca file CSV
df = pd.read_csv(csv_path, parse_dates=['Tanggal'])

# Buat rentang tanggal yang seharusnya ada
start_date = pd.to_datetime("2023-01-01")
end_date = pd.to_datetime("2024-12-31")
expected_dates = pd.date_range(start=start_date, end=end_date, freq='D')

# Ambil tanggal yang ada di file
available_dates = pd.to_datetime(df['Tanggal'].dropna().unique())

# Cari tanggal yang hilang
missing_dates = expected_dates.difference(available_dates)

# Tampilkan hasil
print(f"Jumlah tanggal yang hilang: {len(missing_dates)}")
print("Daftar tanggal yang hilang:")
print(missing_dates)
```
